```python
import google.generativeai as genai
from google.generativeai import types
import os

# Ensure you have your API key set as an environment variable or replace "YOUR_API_KEY"
# genai.configure(api_key=os.environ.get("GOOGLE_API_KEY"))

model = genai.GenerativeModel('gemini-2.5-flash-preview-tts')

prompt_text = "Hello, this is a test of the Gemini 2.5 Flash Text-to-Speech model."

response = model.generate_content(
    contents=[{"parts": [{"text": prompt_text}]}],
    safety_settings={'HARASSMENT': 'BLOCK_NONE'}, # Adjust safety settings as needed
    generation_config=types.GenerateContentConfig(
        response_modalities=['AUDIO'],
        speech_config=types.SpeechConfig(
            # You can specify a prebuilt voice or other voice settings here
            # Example using a prebuilt voice:
            prebuilt_voice_config=types.PrebuiltVoiceConfig(voice_name='Kore'),
            # For multi-speaker:
            # multi_speaker_voice_config=types.MultiSpeakerVoiceConfig(...)
        )
    )
)

# The audio content will be in response.candidates[0].content.parts[0].inline_data.data
# You'll need to save this base64 encoded data to an audio file (e.g., .wav)
if response.candidates:
    audio_data = response.candidates[0].content.parts[0].inline_data.data
    # For example, to save as a WAV file:
    with open("output.wav", "wb") as f:
        f.write(base64.b64decode(audio_data))
    print("Audio data generated (base64 encoded). You'll need to decode and save it.")
else:
    print("No audio data generated.")
```

```python
import google.generativeai as genai
from google.generativeai import types
import wave
import base64 # Import base64 to decode the audio data

# Set up the wave file to save the output:
def wave_file(filename, pcm_data, channels=1, rate=24000, sample_width=2):
    # pcm_data is expected to be bytes, if it comes as base64 encoded string, it needs decoding
    if isinstance(pcm_data, str):
        pcm_data = base64.b64decode(pcm_data)

    with wave.open(filename, "wb") as wf:
        wf.setnchannels(channels)
        wf.setsampwidth(sample_width)
        wf.setframerate(rate)
        wf.writeframes(pcm_data)

# --- FIX START ---
# Configure the API key directly
genai.configure(api_key="YOUR_ACTUAL_GEMINI_API_KEY") # <<< IMPORTANT: Replace with your actual API key
# --- FIX END ---

response = genai.generate_content( # <<< Changed from client.models.generate_content
    model="gemini-2.5-flash-preview-tts",
    contents="Say cheerfully: Have a wonderful day!",
    generation_config=types.GenerateContentConfig( # It's `generation_config` for the main call
        response_modalities=["AUDIO"],
        speech_config=types.SpeechConfig(
            voice_config=types.VoiceConfig(
                prebuilt_voice_config=types.PrebuiltVoiceConfig(
                    voice_name='Kore',
                )
            )
        ),
    )
)

# The audio data is usually base64 encoded and needs to be decoded before writing to wave file
data = response.candidates[0].content.parts[0].inline_data.data

file_name='out.wav'
wave_file(file_name, data) # Saves the file to current directory
print(f"Audio saved to {file_name}")
```
